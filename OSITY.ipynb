To train our model, we will experiment two methods to evaluate the outputs.

## Regular Training
This involves following the regular training procedure from Ultralytics. The process involves having general parameters like:
- Epochs: 300
- Image size: 640
- Batch: 4

## Fine-tuning hyperparameters before training
We explored using an automated hyperparamter optimization by combining:
- Bayesian optimization or grid search
- Multiple short training trials
- Evluation to find the best configuration

Several aspects that could be fine tuned during the process include:
- **Learning rate**: Initial and Final
- **Momentum**: Momentum in optimizer
- **Weight Decay**: Regularization to avoid overfitting
- **Warmup epochs**: Momentum to start of training
- **Box**: Box loss gain
- **Class**: Classification loss gain
- **dfl**: Distribution focal loss gain
- **hsv_h, hsv_s, hsv_v**: color augmentations
- **degrees**: Random rotation
- **translate**: Random shift in image
- **scale**: random image scaling
- **shear**: shearing of image
- **fliplr**: horizontal flipping probability
- **mixup, cutmix**: Augmenntation techniques
- **copy_paste**: paste objects from onr image to another

## What does this do?
This process tweaks different training settings (hyperparameters) to find the combination that gives the best model performance, usually in terms of *mAP*

# What happens during training?
During training, the model learns. **But how?**

The model sees input images and their ground truth (masks, boxes, and class labels), learns to make predictions, compares those predictions to the ground truths, claculates a loss (error), and adjusts its  weights (*learned internal parameters*) to reduce this loss via **back propagation**.
So by the end of training, the model **weights** are optimized to recognize the classes as accurately as possible.

# What happens after training?
After training, the model selects the best checkpoints by doing the following automatically:
- last.pt: the final model after all epochs
- best.pt: the one the performed best on validation set, based on mAP@50.
**Note: best.pt is usually used*

The following reports are generated:
- Precision: how many of the model's prediction were correct
- Recall: how mnay actual objects the model detected.
- mAP@50: mean average precidion at 50% IoU (Intersection over Union)
- mAP@50-95: averaged over stricter thresholds.

**These give us a picture of what the model is good at and what it struggles with**.

# Adding counts and percentage to image and video
The goal of the project was to provide a concept that despite the similarity in the physical features of the samples, we can still use computer vision to classify them.

Additionally, the plan is to also **count, then calculate the proportion (%) of each class in an image or video frame**.

```python
# Assign your model location to model
model = YOLO('/content/runs/segment/train/weights/best.pt')
# Assign the location of the inference file to source
source = '/content/drive/MyDrive/Env/Particle Classifier/Videos/New/20250725_145736000_iOS.MOV'
```

```python
import cv2
from ultralytics import solutions

region_points = [(2500, 20), (2500, 4000)]
counter = solutions.ObjectCounter(show=True, region=region_points,
                                  model=model)

if isinstance(source, str) and (source.endswith(".jpg") or source.endswith(".png")):
    im0 = cv2.imread(source)
    results = counter(im0)
    print("Class-wise counts:", results.classwise_count)
    cv2.imwrite("output_image.jpg", results.plot_im)
else:
    cap = cv2.VideoCapture(source)
    assert cap.isOpened(), "Error reading video file or stream"
    w, h, fps = (int
     (cap.get(x)) for x in
      (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))
    video_writer = cv2.VideoWriter("output_video.avi",
                                   cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))
    while cap.isOpened():
        success, im0 = cap.read()
        if not success:
            break
        results = counter(im0)
        print("Class-wise counts:", results.classwise_count)
        video_writer.write(results.plot_im)
    cap.release()
    video_writer.release()
```
